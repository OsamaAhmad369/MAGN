```python
import numpy as np
import pandas as pd

ca_meta = pd.read_csv('../ca_meta.csv')
gba_meta = ca_meta[ca_meta.District == 4]
gba_meta = gba_meta.reset_index()
gba_meta = gba_meta.drop(columns=['index'])
gba_meta.to_csv('gba_meta.csv', index=False)
print(gba_meta[gba_meta.duplicated(subset=['Lat', 'Lng'])])
gba_meta
```

```python


# Load the main CSV file
main_df = pd.read_csv('../ca_meta.csv')

# Load the subset CSV files
subset1_df = pd.read_csv('gba_meta.csv')
subset2_df = pd.read_csv('../gla/gla_meta.csv')
subset3_df = pd.read_csv('../sd/sd_meta.csv')

# Combine all subsets into one DataFrame
all_subsets = pd.concat([subset1_df, subset2_df, subset3_df])

# Drop duplicates from the combined DataFrame
all_subsets = all_subsets.drop_duplicates()

# Filter out the rows from main_df that are present in all_subsets
# Assuming 'ID' is the column to compare
filtered_main_df = main_df[~main_df['ID'].isin(all_subsets['ID'])]

# Save the filtered data to a new CSV file
filtered_main_df.to_csv('filtered_main_csv.csv', index=False)
```

```python
# code for checking adj stat
gba_rn_adj = np.load('gba_rn_adj.npy')
node_num = gba_rn_adj.shape[0]

print(gba_rn_adj[0,0])
gba_rn_adj[np.arange(node_num), np.arange(node_num)] = 0
print(gba_rn_adj[0,0])

print('edge number', np.count_nonzero(gba_rn_adj))
print('node degree', np.mean(np.count_nonzero(gba_rn_adj, axis=-1)))
print('sparsity', np.count_nonzero(gba_rn_adj) / (node_num**2) * 100)
```
